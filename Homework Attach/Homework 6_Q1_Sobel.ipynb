```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
### change to your image
img = cv2.imread('Yue_Image_Upload.jpg')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# convert to grayscale
gray = 0.299*img_rgb[:,:,0] + 0.587*img_rgb[:,:,1] + 0.114*img_rgb[:,:,2]
gray = gray.astype(np.float64)

plt.imshow(gray, cmap='gray')
plt.title("Grayscale")
plt.axis('off')
```

```python
# Sobel kernels to be complete
Sx = np.array([
    [-1, 0, 1],
    [?, ?, ?],
    [?, ?, ?]
])

Sy = np.array([
    [-1, -2, -1],
    [ ?,  ?,  ?],
    [ ?,  ?,  ?]
])
```

```python
def convolve2d(image, kernel):
    h, w = image.shape
    kh, kw = kernel.shape

    pad_h = kh // 2
    pad_w = kw // 2

    padded = np.pad(image, ((pad_h, pad_h), (pad_w, pad_w)), mode='constant')

    output = np.zeros_like(image)

    # convolution
    for i in range(h):
        for j in range(w):
            region = padded[i:i+kh, j:j+kw]
            output[i, j] = np.sum(region * kernel)

    return output

Gx = convolve2d(gray, Sx)
Gy = convolve2d(gray, Sy)
G = np.sqrt(Gx**2 + Gy**2)
G_disp  = G / np.max(G)
```

```python
plt.figure()
plt.imshow(G_disp, cmap='gray')
plt.axis('off')

plt.show()
```
